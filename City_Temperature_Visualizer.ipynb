```python
import pandas as pd
from google.colab import drive
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import sys

drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/2.city_temperature.csv')

print('Hello sir, welcome!')
print('World temperature database query')

bölge = input('Please enter region= ')
ülke = input('Please enter country= ')
şehir = input('Please enter city= ')
tarih_str = input('Please enter date (YYYY-MM-DD format)= ')

try:
    tarih = datetime.strptime(tarih_str, '%Y-%m-%d')
    yıl, ay, gün = tarih.year, tarih.month, tarih.day
except ValueError:
    print("Invalid date format. Please use YYYY-MM-DD. Exiting.")
    sys.exit()

conclusion = df[
    (df['Country'].astype(str).str.lower() == ülke.lower()) &
    (df['City'].astype(str).str.lower() == şehir.lower()) &
    (df['Year'] == yıl) &
    (df['Month'] == ay) &
    (df['Day'] == gün) &
    (df['Region'].astype(str).str.lower() == bölge.lower())
]

print("\n--- Search result for the specific date ---")
if conclusion.empty:
    print("There are no results you are looking for.")
else:
    print(conclusion)
    row = conclusion.iloc[0]
    avg_temp = row['AvgTemperature']

    if avg_temp == -99:
        print("Note: This record has a missing/placeholder temperature value (-99) in the dataset. No single-day chart will be drawn.")
    else:
        label = f"{row['City']} ({row['Year']}-{row['Month']:02d}-{row['Day']:02d})"
        plt.figure(figsize=(8, 6))
        sns.barplot(x=[label], y=[avg_temp], palette="viridis")
        plt.title(f"{label} Average Temperature")
        plt.ylabel('Average Temperature')
        plt.xlabel('City and Date')
        plt.show()

print("\n--- Generating Monthly Temperature Trend for the Year ---")

yearly_data = df[
    (df['Country'].astype(str).str.lower() == ülke.lower()) &
    (df['City'].astype(str).str.lower() == şehir.lower()) &
    (df['Year'] == yıl) &
    (df['Region'].astype(str).str.lower() == bölge.lower()) &
    (df['AvgTemperature'] != -99)
]

if yearly_data.empty:
    print("Not enough valid data to plot a yearly trend.")
else:
    monthly_avg = yearly_data.groupby('Month')['AvgTemperature'].mean().reset_index()

    plt.figure(figsize=(10, 6))
    sns.lineplot(data=monthly_avg, x='Month', y='AvgTemperature', marker='o', color='coral', linewidth=2.5)

    plt.title(f"{şehir.title()} - Monthly Average Temperatures in {yıl}")
    plt.xlabel('Month')
    plt.ylabel('Average Temperature')
    plt.xticks(range(1, 13))
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()
```
